## Produktive Umgebung

**Produktive Umgebung:** Die produktive Umgebung stellt Backstage für den regulären Betrieb bereit und sollte mit einer sicheren Authentifizierung sowie einer stabilen Infrastruktur betrieben werden.

```python
%%bash
cat <<'EOF' | tee ~/mybackstage/app-config.production.yaml 
app:
  title: ${BACKSTAGE_NAME}
  baseUrl: http://${BACKSTAGE_HOSTNAME}:${BACKSTAGE_PORT}

organization:
  name: ${BACKSTAGE_NAME}

backend:
  baseUrl: http://${BACKSTAGE_HOSTNAME}:7007
  listen:
    host: 0.0.0.0
    port: 7007

  cors:
    origin: http://${BACKSTAGE_HOSTNAME}:${BACKSTAGE_PORT}
    methods: [GET, HEAD, PATCH, POST, PUT, DELETE]
    credentials: true

# in Produktiven Umgebungen entfernen!
auth:
  providers:
    guest:
      dangerouslyAllowOutsideDevelopment: true

catalog:
  import:
    entityFilename: catalog-info.yaml
    pullRequestBranchName: backstage-integration
  rules:
    - allow: [Component, System, API, Resource, Location, Templates]
  locations:
    # Auto Shop GmbH        
    - type: url
      target: https://gitlab.com/ch-mc-b/autoshop-ms/infra/backstage/-/blob/main/org.yaml
      rules:
        - allow: [User, Group, Template]   
EOF
```

**Konfiguration prüfen:** Mit `yarn backstage-cli config:print` wird die zusammengeführte und aufgelöste Backstage-Konfiguration ausgegeben, ohne die Anwendung zu starten.


```python
%%bash
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Produktion"
export BACKSTAGE_PORT="3002"

source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn backstage-cli config:print --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.production.yaml
```

**Backstage starten:** Backstage wird mit den gewünschten Konfigurationsdateien gestartet.

```python
%%bash
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Produktion"
export BACKSTAGE_PORT="3002"

echo "http://$(cat ~/data/server-ip):${BACKSTAGE_PORT}"
source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn start --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.production.yaml
```
