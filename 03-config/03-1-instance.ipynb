## Neue Backstage Instanz

Für die nachfolgenden Übungen richten wir eine eigene Backstage-Instanz ein.


```python
%%bash
source ~/.nvm/nvm.sh
rm -rf ~/mybackstage
npx @backstage/create-app@latest --path ~/mybackstage <<<backstage
```

Falls die Installation zuvor mit einer Fehlermeldung wie `YN0016 ([...] quarantined)` abgebrochen ist, blockiert das Yarn 4 Sicherheitsfeature `npmMinimalAgeGate` brandneue npm-Pakete für 24 Stunden zum Schutz vor Schadsoftware. 

Führe in diesem Fall bitte die nachfolgenden Code-Zeilen aus, um dieses Alterslimit vorübergehend zu deaktivieren und die Installation erfolgreich abzuschliessen.


```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage
# Das Alterslimit für npm-Pakete in diesem Projekt auf 0 setzen
yarn config set npmMinimalAgeGate 0
# Die Installation manuell und erfolgreich durchführen
yarn install
```

### Polyfill Patch

Ein **Polyfill-Patch** ergänzt fehlende Browser-Funktionen per Code, damit eine Anwendung trotzdem läuft.
In deinem Fall wird `crypto.randomUUID()` nur dann nachgerüstet, wenn der Browser diese Funktion nicht bereits bereitstellt.

Backstage braucht `crypto.randomUUID()` hier im Signals-Plugin, um eindeutige IDs für Signal-Subscriptions im Frontend zu erzeugen.


```python
%%bash
cd ~/mybackstage || exit 1
f=packages/app/src/index.tsx

grep -q "crypto.randomUUID" "$f" && echo "Schon vorhanden" && exit 0

cp "$f" "$f.bak"
sed -i '1i\
if (globalThis.crypto && !globalThis.crypto.randomUUID) {\
  globalThis.crypto.randomUUID = (() => crypto.getRandomValues(new Uint8Array(16)).join("-")) as Crypto["randomUUID"];\
}\
' "$f"
```
