## Testumgebung

**Testumgebung:** Die Testumgebung dient dazu, Änderungen und Konfigurationen sicher zu prüfen, bevor sie in die produktive Umgebung übernommen werden.

```python
%%bash
cat <<'EOF' | tee ~/mybackstage/app-config.test.yaml 
app:
  title: ${BACKSTAGE_NAME}
  baseUrl: http://${BACKSTAGE_HOSTNAME}:${BACKSTAGE_PORT}

organization:
  name: ${BACKSTAGE_NAME}

backend:
  baseUrl: http://${BACKSTAGE_HOSTNAME}:7007
  listen:
    host: 0.0.0.0
    port: 7007

  cors:
    origin: http://${BACKSTAGE_HOSTNAME}:${BACKSTAGE_PORT}
    methods: [GET, HEAD, PATCH, POST, PUT, DELETE]
    credentials: true
EOF
```

**Konfiguration prüfen:** Mit `yarn backstage-cli config:print` wird die zusammengeführte und aufgelöste Backstage-Konfiguration ausgegeben, ohne die Anwendung zu starten.

```python
%%bash
set -a
source ~/data/env-platen.py
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Test"
export BACKSTAGE_PORT="3001"

source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn backstage-cli config:print --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.test.yaml
```

**Backstage starten:** Backstage wird mit den gewünschten Konfigurationsdateien gestartet.

```python
%%bash
set -a
source ~/data/env-platen.py
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Test"
export BACKSTAGE_PORT="3001"

echo "http://$(cat ~/data/server-ip):${BACKSTAGE_PORT}"
source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn start --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.test.yaml
```
