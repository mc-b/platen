## Minimales MkDocs Beispiel

Das folgende Beispiel zeigt die kleinste sinnvolle Struktur einer MkDocs-Dokumentation mit Konfiguration und Startseite.


```python
%%bash
mkdir -p minimal-techdocs/docs
cd minimal-techdocs

cat > mkdocs.yml <<'EOF'
site_name: Minimal TechDocs

nav:
  - Startseite: index.md
EOF

cat > docs/index.md <<'EOF'
# Minimal TechDocs

Dies ist eine minimale MkDocs-Dokumentation.

## Inhalt

Die Dokumentation wird als Markdown-Datei gepflegt und kann durch Backstage TechDocs dargestellt werden.
EOF

cat > catalog-info.yaml <<'EOF'
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: minimal-techdocs
  title: Minimal TechDocs
  description: Minimales Beispiel für MkDocs und Backstage TechDocs
  annotations:
    backstage.io/techdocs-ref: dir:.
spec:
  type: documentation
  lifecycle: experimental
  owner: user:default/guest
EOF
```

**Installation von MkDocs:** MkDocs wird in einer virtuellen Python-Umgebung installiert, damit die benötigten Pakete isoliert vom restlichen System verwaltet werden.

```python
%%bash
python3 -m venv ~/.techdocs
source ~/.techdocs/bin/activate
pip install mkdocs
```

**Start der lokalen MkDocs-Instanz:** Die lokale MkDocs-Instanz wird auf `0.0.0.0` gestartet, damit die Dokumentation auch von anderen Geräten im gleichen Netzwerk erreichbar ist.

```python
%%bash
echo "http://$(cat ~/data/server-ip):8000"
cd minimal-techdocs
mkdocs serve --dev-addr=0.0.0.0:8000
```
