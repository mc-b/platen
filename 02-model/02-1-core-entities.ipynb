## Kernelemente (Core Entities)

Für unser neues Projekt "Lemon Shop" wollen wir folgendes System erstellen

    lemon-shop (System)
    │
    ├── shop-frontend
    │   └── shop-api
    │
    ├── shop-api
    │   ├── product-service
    │   ├── order-service
    │   └── payment-service
    │
    ├── product-service
    │
    ├── order-service
    │   ├── product-service
    │   └── payment-service
    │
    └── payment-service
        


```python
%%bash
mkdir ~/lemon-shop
cat <<EOF >~/lemon-shop/entities.yaml
apiVersion: backstage.io/v1alpha1
kind: System
metadata:
  name: lemon-shop
  namespace: default
  title: Lemon Shop
  description: Übergeordnetes System für die Lemon-Shop-Applikation
  tags:
    - lemon-shop
    - ecommerce
spec:
  owner: user:default/guest
  lifecycle: experimental

---
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: shop-frontend
  namespace: default
  title: Shop Frontend
  description: Benutzeroberfläche des Lemon Shops
  tags:
    - frontend
    - web
    - lemon-shop
spec:
  type: website
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
  dependsOn:
    - component:default/shop-api

---
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: shop-api
  namespace: default
  title: Shop API
  description: Zentrale API und Einstiegspunkt für das Shop Frontend
  tags:
    - python
    - api
    - backend
    - lemon-shop
spec:
  type: service
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
  dependsOn:
    - component:default/product-service
    - component:default/order-service
    - component:default/payment-service

---
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: product-service
  namespace: default
  title: Product Service
  description: Verwaltet Produkte, Preise und Produktinformationen
  tags:
    - python
    - service
    - products
    - lemon-shop
spec:
  type: service
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop

---
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: order-service
  namespace: default
  title: Order Service
  description: Verwaltet Warenkörbe und Bestellungen
  tags:
    - python
    - service
    - orders
    - lemon-shop
spec:
  type: service
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
  dependsOn:
    - component:default/product-service
    - component:default/payment-service

---
apiVersion: backstage.io/v1alpha1
kind: Component
metadata:
  name: payment-service
  namespace: default
  title: Payment Service
  description: Verarbeitet Zahlungsvorgänge für Bestellungen
  tags:
    - python
    - service
    - payments
    - lemon-shop
spec:
  type: service
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
EOF
```
