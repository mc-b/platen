## Organizational Entities

```python
%%bash
cat <<EOF >~/lemon-shop/examples/org.yaml
apiVersion: backstage.io/v1alpha1
kind: Group
metadata:
  name: shop-team
  description: Webshop der Lemon Shop GmbH
  links:
    - url: https://gitlab.com/ch-mc-b/lemonshop-ms/app/shop
      title: Sourcecode
      icon: git  
    - url: https://gitlab.com/ch-mc-b/lemonshop-ms/app/shop/container_registry
      title: Container Registry
      icon: registry 
spec:
  type: team
  parent: app
  children: []
  members:
    - anna.lehmann
    - david.kraus

---
apiVersion: backstage.io/v1alpha1
kind: User
metadata:
  name: anna.lehmann
spec:
  profile:
    displayName: Anna Lehmann
    email: anna.lehmann@lemonshopgmbh.ch
  memberOf: [shop-team]

---
apiVersion: backstage.io/v1alpha1
kind: User
metadata:
  name: david.kraus
spec:
  profile:
    displayName: David Kraus
    email: david.kraus@lemonshopgmbh.ch
  memberOf: [shop-team]
EOF
```

```python
%%bash
cat <<EOF >>~/lemon-shop/examples/org.yaml
---
apiVersion: backstage.io/v1alpha1
kind: Group
metadata:
  name: backoffice-team
  description: Backoffice Team der Lemon Shop GmbH
  links:
    - url: https://gitlab.com/ch-mc-b/lemonshop-ms/app/backoffice
      title: Sourcecode
      icon: git  
    - url: https://gitlab.com/ch-mc-b/lemonshop-ms/app/backoffice/container_registry
      title: Container Registry
      icon: registry   
spec:
  type: team
  parent: app
  children: []
---
apiVersion: backstage.io/v1alpha1
kind: User
metadata:
  name: lisa.mueller
spec:
  profile:
    displayName: Lisa Müller
    email: lisa.mueller@lemonshopgmbh.ch
  memberOf: [backoffice-team]
---
apiVersion: backstage.io/v1alpha1
kind: User
metadata:
  name: thomas.schmidt
spec:
  profile:
    displayName: Thomas Schmidt
    email: thomas.schmidt@lemonshopgmbh.ch
  memberOf: [backoffice-team]
EOF
```
