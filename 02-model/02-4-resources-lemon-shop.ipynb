## Resources

Die beiden Resource-Entities bilden die technische Infrastruktur des Lemon Shops ab. `database.yaml` beschreibt die PostgreSQL-Datenbank für Produkte, Bestellungen und Zahlungen, während `kubernetes.yaml` die Laufzeitumgebung der Applikationen darstellt.

Durch die Zuordnung zum System `lemon-shop` können Components ihre Abhängigkeiten zu diesen Ressourcen im Backstage-Katalog sichtbar machen.


```python
%%bash
cat > ~/lemon-shop/examples/database.yaml <<'EOF'
apiVersion: backstage.io/v1alpha1
kind: Resource
metadata:
  name: database
  namespace: default
  title: Database
  description: PostgreSQL-Datenbank für Produkte, Bestellungen und Zahlungen
  tags:
    - lemon-shop
    - postgresql
    - database
spec:
  type: database
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
EOF

cat > ~/lemon-shop/examples/kubernetes.yaml <<'EOF'
apiVersion: backstage.io/v1alpha1
kind: Resource
metadata:
  name: kubernetes
  namespace: default
  title: Kubernetes
  description: Kubernetes-Laufzeitumgebung für die Applikationen
  tags:
    - lemon-shop
    - kubernetes
    - platform
spec:
  type: kubernetes-cluster
  lifecycle: experimental
  owner: user:default/guest
  system: lemon-shop
EOF
```
