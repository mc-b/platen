## Frontend-Plugin mit Menüpunkt

Backstage erzeugt das Plugin mit dem offiziellen CLI-Template. 

Anschliessend wird die Legacy-Verdrahtung durch eine `PageBlueprint`-Extension ersetzt. 

Die `PageBlueprint`-Extension definiert eine eigenständige Seite mit Route, Titel, Icon und Ladefunktion. Im neuen Frontend-System wird daraus automatisch ein aufrufbarer Menüpunkt in der Backstage-Navigation.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/
yarn new --select frontend-plugin --option pluginId=hello
```

Der Befehl installiert die benötigten Abhängigkeiten, damit das Plugin als Backstage-Frontend-Plugin registriert und mit einem Icon im Menü dargestellt werden kann.


```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/
yarn --cwd plugins/hello add   @backstage/frontend-plugin-api   @remixicon/react
```

Eine Page erzeugt im neuen Frontend-System automatisch einen Menüpunkt.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/

mkdir -p plugins/hello/src/components/HelloPage

cat > plugins/hello/src/components/HelloPage/index.tsx <<'EOF'
import React from 'react';

export function HelloPage() {
  return (
    <main style={{ padding: 24 }}>
      <h1>Hello Plugin</h1>
      <p>Diese Seite stammt aus einem eigenen Backstage-Plugin.</p>
    </main>
  );
}
EOF
```

Die Route beschreibt den Navigationspfad zur Seite, während das Plugin die zugehörigen Erweiterungen und Funktionen als Backstage-Modul bündelt.


```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/

cat > plugins/hello/src/routes.ts <<'EOF'
import { createRouteRef } from '@backstage/frontend-plugin-api';

export const rootRouteRef = createRouteRef();
EOF

cat > plugins/hello/src/plugin.tsx <<'EOF'
import React from 'react';
import {
  createFrontendPlugin,
  PageBlueprint,
} from '@backstage/frontend-plugin-api';
import { RiPuzzleLine } from '@remixicon/react';
import { rootRouteRef } from './routes';

const helloPage = PageBlueprint.make({
  params: {
    routeRef: rootRouteRef,
    path: '/hello',
    title: 'Hello',
    icon: <RiPuzzleLine />,
    loader: () =>
      import('./components/HelloPage').then(m => <m.HelloPage />),
  },
});

export const helloPlugin = createFrontendPlugin({
  pluginId: 'hello',
  extensions: [helloPage],
  routes: {
    root: rootRouteRef,
  },
});
EOF

cat > plugins/hello/src/index.ts <<'EOF'
export { helloPlugin as default } from './plugin';
EOF
```

**Backstage starten:** Backstage wird mit dem neuen Plugin gestartet.

```python
%%bash
set -a
source ~/data/env-platen.py
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Produktion"
export BACKSTAGE_PORT="3002"

echo "http://$(cat ~/data/server-ip):${BACKSTAGE_PORT}"
source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn start --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.production.yaml
```
