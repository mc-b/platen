## Catalog-Plugin mit Component-Tab

Das Plugin wird mit dem offiziellen CLI-Template erzeugt. 

Eine `EntityContentBlueprint`-Extension ergänzt Catalog-Entities um einen Tab.

Der Filter beschränkt den Tab auf `Component`-Entities.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/
yarn new --select frontend-plugin --option pluginId=component-info
```

Der Befehl installiert die benötigten Abhängigkeiten.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/
yarn --cwd plugins/component-info add   @backstage/frontend-plugin-api   @backstage/plugin-catalog-react
```

Inhalt des Tabs.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/

mkdir -p plugins/component-info/src/components/ComponentInfo

cat > plugins/component-info/src/components/ComponentInfo/index.tsx <<'EOF'
import React from 'react';
import { useEntity } from '@backstage/plugin-catalog-react';

export function ComponentInfo() {
  const { entity } = useEntity();

  return (
    <main style={{ padding: 24 }}>
      <h2>{entity.metadata.title ?? entity.metadata.name}</h2>
      <p>Typ: {entity.spec?.type as string}</p>
      <p>Lifecycle: {entity.spec?.lifecycle as string}</p>
      <p>Owner: {entity.spec?.owner as string}</p>
    </main>
  );
}
EOF
```

Catalog-Tab gemäss offizieller `EntityContentBlueprint`-Anleitung.

```python
%%bash
source ~/.nvm/nvm.sh
cd ~/mybackstage/

cat > plugins/component-info/src/plugin.tsx <<'EOF'
import React from 'react';
import { createFrontendPlugin } from '@backstage/frontend-plugin-api';
import { EntityContentBlueprint } from '@backstage/plugin-catalog-react/alpha';

const componentInfoContent = EntityContentBlueprint.make({
  params: {
    path: 'component-info',
    title: 'Component Info',
    filter: 'kind:component',
    loader: () =>
      import('./components/ComponentInfo').then(m => <m.ComponentInfo />),
  },
});

export const componentInfoPlugin = createFrontendPlugin({
  pluginId: 'component-info',
  extensions: [componentInfoContent],
});
EOF

cat > plugins/component-info/src/index.ts <<'EOF'
export { componentInfoPlugin as default } from './plugin';
EOF
```

**Backstage starten:** Backstage wird mit dem neuen Plugin gestartet.

```python
%%bash
set -a
source ~/data/env-platen.py
export BACKSTAGE_HOSTNAME=$(cat ~/data/server-ip)
export BACKSTAGE_NAME="Backstage Produktion"
export BACKSTAGE_PORT="3002"

echo "http://$(cat ~/data/server-ip):${BACKSTAGE_PORT}"
source ~/.nvm/nvm.sh
cd ~/mybackstage
yarn start --config ~/mybackstage/app-config.yaml --config ~/mybackstage/app-config.production.yaml
```
